# Agentic Healthcare Chatbot using DeepSeek
This notebook demonstrates how to build a basic agentic healthcare chatbot using the DeepSeek API. It includes step-by-step outputs as requested.

```python
import os
from openai import OpenAI

# Print step start
print('[Step 1] Initializing DeepSeek Client...')

# You can get your API key from DeepSeek platform
# We use the OpenAI SDK since DeepSeek's API is OpenAI-compatible
DEEPSEEK_API_KEY = os.environ.get('DEEPSEEK_API_KEY', 'your_api_key_here')

client = OpenAI(
    api_key=DEEPSEEK_API_KEY,
    base_url='https://api.deepseek.com/v1'
)

print('[Output] Client initialized successfully.')
```

### Define the Healthcare Agentic System Prompt
We define the persona and constraints for the chatbot.

```python
print('[Step 2] Defining System Prompt and Persona...')

system_prompt = """
You are a helpful, empathetic, and highly knowledgeable Agentic Healthcare Chatbot.
Your goal is to assist users with their medical inquiries by providing accurate, well-structured, and easily understandable information.
Always remind the user that you are an AI and not a substitute for professional medical advice, diagnosis, or treatment.
If a user indicates a life-threatening emergency, immediately advise them to contact emergency services.
"""

print('[Output] System prompt defined:')
print(system_prompt.strip())
```

### Provide a User Query and Prepare Messages

```python
print('[Step 3] Preparing conversation messages...')

user_query = "I have been experiencing a mild headache and fatigue for the past two days. What could be the possible reasons, and what should I do?"

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_query}
]

print(f'[Output] User Query: {user_query}')
print(f'[Output] Total messages prepared: {len(messages)}')
```

### Call DeepSeek API to Generate Response

```python
print('[Step 4] Calling DeepSeek API...')

try:
    # deepseek-chat is the standard text model, deepseek-coder is for coding
    response = client.chat.completions.create(
        model="deepseek-chat",
        messages=messages,
        max_tokens=500,
        temperature=0.3  # Lower temperature for more deterministic/factual responses
    )
    
    print('[Output] API Call Successful!')
except Exception as e:
    print(f'[Output] Error during API call: {e}')
    response = None
```

### Process and Print the Final Response

```python
print('[Step 5] Processing and Printing Final Response...')

if response:
    agent_reply = response.choices[0].message.content
    print("\n" + "="*50)
    print("⚕️ HEALTHCARE AGENT RESPONSE ⚕️")
    print("="*50)
    print(agent_reply)
    print("="*50)
else:
    print("No response to display due to an earlier error.")
```
